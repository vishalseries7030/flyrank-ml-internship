## 1. Distributions

*Look before deciding: distributions of your key fields. Note the heavy tails.*

Distribution findings

The content metadata and March 2026 GSC performance were joined using client_hash_id and content_hash_id. The joined table contains 519,606 content rows, of which 176,738 have matched GSC performance groups.

The distributions are highly skewed. Backlinks are zero for 77.85% of non-missing observations, while search volume is zero for 43.41%. Both fields have large maximum values relative to their medians, particularly backlinks (maximum 4,269,360) and search volume (maximum 368,000).

GSC clicks are zero for 61.05% of matched performance groups. Missing GSC matches must not be interpreted as zero impressions or clicks. These distributions suggest that raw counts may be dominated by a small number of high-volume observations, so subsequent signal tests should use robust comparisons and clearly defined eligibility rules.

```python
from huggingface_hub import hf_hub_download
import pyarrow.parquet as pq

repo = "FlyRank/internship-warehouse"

daily_path = hf_hub_download(
    repo_id=repo,
    repo_type="dataset",
    filename="fact_content_daily_performance/month=2026-03/data_0.parquet"
)

content_path = hf_hub_download(
    repo_id=repo,
    repo_type="dataset",
    filename="dim_content.parquet"
)

print("March dataset path:", daily_path)
print("Content dataset path:", content_path)
print("Dataset paths ready!")
```

```python

import pandas as pd
import pyarrow.parquet as pq
from IPython.display import display

# Read only the required content metadata columns
keys = ["client_hash_id", "content_hash_id"]

content_cols = keys + [
    "content_type",
    "word_count",
    "backlinks",
    "search_volume",
    "is_published",
    "is_deleted",
    "optimization_eligible_date",
]

content = pq.read_table(
    content_path,
    columns=content_cols
).to_pandas()

# Aggregate March performance in batches to limit RAM usage
daily_pf = pq.ParquetFile(daily_path)
perf_cols = keys + [
    "report_date",
    "gsc_data_available",
    "gsc_impressions",
    "gsc_clicks",
]

parts = []

for batch in daily_pf.iter_batches(
    batch_size=100_000,
    columns=perf_cols
):
    df = batch.to_pandas()

    # Only use rows where GSC data is explicitly available
    df = df.loc[df["gsc_data_available"].eq(True)]

    if df.empty:
        continue

    grouped = df.groupby(keys).agg(
        gsc_impressions=(
            "gsc_impressions",
            lambda s: s.sum(min_count=1)
        ),
        gsc_clicks=(
            "gsc_clicks",
            lambda s: s.sum(min_count=1)
        ),
        available_rows=("report_date", "size"),
    ).reset_index()

    parts.append(grouped)

if not parts:
    raise ValueError("No rows with available GSC data were found.")

perf = pd.concat(parts, ignore_index=True)

# Combine repeated groups across batches
perf = perf.groupby(keys).agg(
    gsc_impressions=(
        "gsc_impressions",
        lambda s: s.sum(min_count=1)
    ),
    gsc_clicks=(
        "gsc_clicks",
        lambda s: s.sum(min_count=1)
    ),
    available_rows=("available_rows", "sum"),
).reset_index()

# Confirm join keys are unique
assert not content.duplicated(keys).any()
assert not perf.duplicated(keys).any()

audit = content.merge(
    perf,
    on=keys,
    how="left",
    validate="one_to_one"
)

print("Content rows:", len(content))
print("Performance groups with GSC data:", len(perf))
print("Joined rows:", len(audit))
print("Rows without matched GSC performance:", audit["gsc_impressions"].isna().sum())

# Describe heavy-tailed distributions
fields = [
    "word_count",
    "backlinks",
    "search_volume",
    "gsc_impressions",
    "gsc_clicks",
]

rows = []

for field in fields:
    s = pd.to_numeric(audit[field], errors="coerce").dropna()

    rows.append({
        "field": field,
        "non_missing_n": len(s),
        "zero_pct": round((s.eq(0).mean() * 100), 2),
        "median": s.quantile(0.50),
        "p75": s.quantile(0.75),
        "p90": s.quantile(0.90),
        "p95": s.quantile(0.95),
        "p99": s.quantile(0.99),
        "max": s.max(),
    })

distribution_table = pd.DataFrame(rows)
display(distribution_table.round(2))
```

## 2. Signal test #1 / #2 / #3 (verdict each)

*Three safe signals, each with a mini-test and a verdict: CONFIRMED / OPPOSITE / MIXED / FALSE.*

Signal test #1 — Backlinks and GSC impressions

Hypothesis: Content with at least one backlink has higher GSC impressions than content with zero backlinks.

Method: Compared March 2026 content-level GSC impressions for eligible records with non-missing backlink counts and matched GSC performance. Compared group medians and means.

Results: The zero-backlink group contained 82,138 records, while the nonzero-backlink group contained 24,437. Both groups had the same median impressions of 210. Mean impressions were 1,784.82 for zero-backlink content and 1,591.16 for content with at least one backlink.

Verdict: MIXED. The hypothesis is not supported by the median comparison; the mean is actually higher in the zero-backlink group. Heavy-tailed impressions may influence the means, and this unadjusted comparison does not establish causality.

Limitation: This is an observational comparison and does not control for content age, content type, search demand, or other confounding factors.


Signal test #2 — Search volume and GSC impressions

Hypothesis: Content with positive search volume has higher GSC impressions than content with zero search volume.

Method: Compared March 2026 GSC impressions between content with positive and zero search volume, using records with non-missing search volume and matched GSC performance.

Results: The positive-search-volume group contained 97,550 records and had median impressions of 239. The zero-search-volume group contained 63,256 records and had median impressions of 206. Mean impressions were 1,675.56 and 1,820.91, respectively.

Verdict: CONFIRMED (directional). The median comparison supports the hypothesis, although the mean is higher in the zero-volume group. The result is observational and does not establish causality.

Limitation: Search volume is only one possible factor associated with impressions; content type, age, and other factors were not controlled.


Signal Test #3 — Word Count and GSC Impressions

Hypothesis: Content with 1,000 or more words has higher GSC impressions than content with fewer than 1,000 words.

Method: Compared March 2026 GSC impressions for content with matched GSC performance and non-missing word counts. Content was divided into two groups: under 1,000 words and 1,000+ words.

Results:

1,000+ words: 110,985 records; median impressions = 338; mean = 2,142.34.

Under 1,000 words: 10,438 records; median impressions = 2; mean = 107.73.

Verdict: CONFIRMED — directional support. The 1,000+ word group has a substantially higher median impression count.

Limitation: This is an observational comparison. It does not establish that increasing word count causes higher impressions; content age, search demand, content type, and other factors may influence the result.

```python
import pandas as pd
from IPython.display import display

# Compare GSC impressions for shorter vs longer content.
signal3 = audit.loc[
    audit["gsc_impressions"].notna()
    & audit["word_count"].notna()
].copy()

signal3["word_count"] = pd.to_numeric(
    signal3["word_count"], errors="coerce"
)
signal3["gsc_impressions"] = pd.to_numeric(
    signal3["gsc_impressions"], errors="coerce"
)

signal3 = signal3.dropna(
    subset=["word_count", "gsc_impressions"]
)

signal3["word_count_group"] = signal3["word_count"].apply(
    lambda x: "Under 1000 words" if x < 1000
    else "1000+ words"
)

summary3 = signal3.groupby("word_count_group").agg(
    content_count=("gsc_impressions", "size"),
    median_impressions=("gsc_impressions", "median"),
    mean_impressions=("gsc_impressions", "mean"),
).reset_index()

display(summary3.round(2))

medians3 = summary3.set_index(
    "word_count_group"
)["median_impressions"]

short_median = medians3.get("Under 1000 words")
long_median = medians3.get("1000+ words")

print("Under 1000 words median:", short_median)
print("1000+ words median:", long_median)

if pd.isna(short_median) or pd.isna(long_median):
    print("Verdict: FALSE — unable to compare both groups")
elif long_median > short_median:
    print("Verdict: CONFIRMED — 1000+ word group has higher median impressions")
elif long_median < short_median:
    print("Verdict: OPPOSITE — under-1000-word group has higher median impressions")
else:
    print("Verdict: MIXED — both groups have the same median")
```

```python
import numpy as np
import pandas as pd
from IPython.display import display

test_df = audit.loc[
    audit["gsc_impressions"].notna()
    & audit["backlinks"].notna()
].copy()

test_df["backlinks"] = pd.to_numeric(
    test_df["backlinks"], errors="coerce"
)
test_df["gsc_impressions"] = pd.to_numeric(
    test_df["gsc_impressions"], errors="coerce"
)

test_df = test_df.dropna(
    subset=["backlinks", "gsc_impressions"]
)

test_df["backlink_group"] = np.where(
    test_df["backlinks"].eq(0),
    "Zero backlinks",
    "At least 1 backlink"
)

summary = test_df.groupby("backlink_group").agg(
    content_count=("gsc_impressions", "size"),
    median_impressions=("gsc_impressions", "median"),
    mean_impressions=("gsc_impressions", "mean"),
).reset_index()

display(summary.round(2))

medians = summary.set_index("backlink_group")["median_impressions"]
zero_median = medians.get("Zero backlinks", np.nan)
nonzero_median = medians.get("At least 1 backlink", np.nan)

print("Zero-backlink median impressions:", zero_median)
print("Nonzero-backlink median impressions:", nonzero_median)

if pd.isna(zero_median) or pd.isna(nonzero_median):
    print("Verdict: FALSE — unable to compare both groups")
elif zero_median == nonzero_median:
    print("Verdict: MIXED — both groups have the same median")
elif nonzero_median > zero_median:
    print("Verdict: CONFIRMED — nonzero-backlink group has a higher median")
else:
    print("Verdict: OPPOSITE — zero-backlink group has a higher median")
```

## 3. The flag-linked test

*Pick a signal one of FlyRank's real flags relies on. Does the data support the rule's assumption?*

Results: The largest group contains 176,568 published, non-deleted records, with median GSC impressions of 174. The other flag combinations contain only 8 and 162 records, so their observed medians and means are much less stable.

Verdict: MIXED / INCONCLUSIVE. The data describes performance among records with matched GSC data, but the highly imbalanced group sizes do not provide strong evidence that the flag rule itself is correct or incorrect. Flag semantics and the production eligibility rule should be verified before making operational changes.

```python
import pandas as pd
from IPython.display import display

# Check whether published/deleted flags relate to observed GSC performance.
flag_df = audit.loc[
    audit["gsc_impressions"].notna()
].copy()

for col in ["is_published", "is_deleted", "gsc_impressions"]:
    flag_df[col] = pd.to_numeric(flag_df[col], errors="coerce")

flag_df = flag_df.dropna(
    subset=["is_published", "is_deleted", "gsc_impressions"]
)

flag_summary = flag_df.groupby(
    ["is_published", "is_deleted"], dropna=False
).agg(
    content_count=("gsc_impressions", "size"),
    median_impressions=("gsc_impressions", "median"),
    mean_impressions=("gsc_impressions", "mean"),
).reset_index()

display(flag_summary.round(2))

print("Note: This is an observational flag audit.")
print("It does not prove that either flag causes a performance difference.")
```

## 4. What this means in practice

*Two or three sentences: what a content team should take from this.*

Content teams should avoid ranking content using raw counts alone because backlinks, search volume, and impressions have highly skewed distributions. Use explicit data-availability checks, robust comparisons, and clear eligibility rules; treat missing GSC matches as unknown rather than zero. Flag-linked differences should be interpreted cautiously when groups are highly imbalanced, and these observational results should support review rather than automatic decisions.
