## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

### My lane as an ML task

**Task type: Ranking / scoring.**

The decision is which content items should be reviewed first when editor capacity is limited. The output would be a priority score that helps an editor rank content items for review. A wrong ranking can waste editor time or cause a potentially important item to be reviewed later. ML can be useful because multiple content and performance signals can interact in ways that are difficult to capture reliably with a single fixed rule.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

### Target or proxy

The output is a **priority score** for content review. The starter dataset does not contain a directly observed future review-priority label, so I will treat the score as a decision-support proxy rather than claim it is an observed outcome. The proxy will combine measurable signals available in the dataset, while excluding `trend_direction` and `trend_pct` because they are derived from trend information. The purpose is to identify which content items should be reviewed first, not to claim that the score predicts a causal outcome.

## 3. Success metric

*One metric you can defend. What number means 'good'?*
### Success metric

### Success metric

The primary evaluation metric will be **precision@K** if an observed priority outcome becomes available for evaluation. Here, K represents the number of content items an editor can realistically review first. The metric would measure the proportion of truly high-priority items within the top K recommendations. In the current starter dataset, there is no direct observed high-priority outcome label, so this metric cannot be calculated yet and the current score should be treated as decision support rather than a validated predictive model.


```python

# Define the evaluation metric for the prioritization task

K = 20

print("Task type: Ranking / scoring")
print("Primary metric: precision@K")
print("K:", K)
print("Current limitation: no observed high-priority outcome label")

```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*
### Why ML beats a fixed rule

A single fixed rule may miss useful combinations of signals. Content priority can depend on factors such as search demand, recent traffic, historical performance, content age, and freshness. These signals can interact, and their importance may differ across content items. A ranking model can learn patterns from multiple signals and produce a consistent priority score, while still being used as decision support rather than an automatic decision-maker.


```python
# Show the main signals available for the ML framing

signal_columns = [
    "search_volume",
    "impressions_90d",
    "clicks_90d",
    "sessions_90d",
    "impressions_last_30d",
    "clicks_last_30d",
    "sessions_last_30d",
    "content_age_days",
    "days_since_last_update"
]

signal_summary = df[signal_columns].describe().T

display(signal_summary)
```
